```python
missing_values = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
})

missing_values.sort_values(
    by="Missing Percentage",
    ascending=False
)
```

Category & Transactions

```python
print("Transaction types:")
display(df["transaction_type"].value_counts())

print("Payment statuses:")
display(df["payment_status"].value_counts())

print("Regions:")
display(df["region"].value_counts())
```

Check Financial Values Consistency

```python
df["calculated_net_amount"] = (
    df["gross_amount_brl"]
    - df["discount_amount_brl"]
    - df["tax_amount_brl"]
)

df["net_amount_difference"] = (
    df["net_amount_brl"]
    - df["calculated_net_amount"]
)

print(
    "Inconsistent financial records:",
    (df["net_amount_difference"].abs() > 0.01).sum()
)
```

Check Date Range

```python
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"]
)

print("Start date:", df["transaction_date"].min())
print("End date:", df["transaction_date"].max())

print("Number of months:",
      df["transaction_date"].dt.to_period("M").nunique())
```
